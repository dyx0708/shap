```python
import shap
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor

plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = 'Times New Roman'
plt.rcParams['font.size'] =14 # 设置字体大小为1

# 训练模型
model_planet = RandomForestRegressor(n_estimators=1000, random_state=0, max_depth=13, min_samples_split=2, min_samples_leaf=1)
model_planet.fit(X_train_planet, y_train_planet)  # 使用 PLANET 数据进行训练
y_pred_planet = model_planet.predict(X_test_planet)
y_train_pred_planet = model_planet.predict(X_train_planet)

# 构建 SHAP 解释器
explainer_planet = shap.TreeExplainer(model_planet)

# 计算测试集的 SHAP 值
# shap_values_planet = explainer_planet.shap_values(X_train_planet)
shap_values_planet = explainer_planet.shap_values(X_train_planet)

# 特征标签
# labels_planet = X_test_planet.columns
labels_planet = X_test_planet.columns

# 调整画布大小和字体大小
plt.figure(figsize=(8, 6), dpi=2000)
shap.summary_plot(shap_values_planet,X_train_planet, feature_names=labels_planet, plot_type="dot", show=False, color=plt.get_cmap("viridis"))

# 设置标题和轴标签的字体大小
plt.title('(a) SHAP Summary Plot for PlanetScope',fontsize=12)
plt.xlabel('SHAP Value',fontsize=12)

# 设置X轴刻度
plt.xlim(-2,2)  # 从 -1 到 +1，每隔 0.2
plt.xticks(ticks=[-2+ i * 0.4 for i in range(11)])  # 从 -1 到 1，每隔 0.2

# 调整特征标签的字体大小
for label in plt.gca().get_xticklabels():
    label.set_fontsize(15)  # 设置 X 轴特征字体大小

for label in plt.gca().get_yticklabels():
    label.set_fontsize(15)  # 设置 Y 轴特征字体大

#设置背景的颜色
plt.gca().set_facecolor('#ECF3FC') 

# 获取当前的 colorbar 并加宽
cbar = plt.gcf().get_axes()[-1]  # 获取 colorbar
    
# plt.savefig(r'E:\9月\实验测试\shap分析\p_plot_median_train_clay.jpg', dpi=2000, bbox_inches='tight')
# 显示图像

plt.show()
```

```python
import numpy as np
import pandas as pd

# 1. 计算每个特征的平均绝对 SHAP 值
mean_abs_shap_planet = np.abs(shap_values_planet).mean(axis=0)

# 2. 构建 DataFrame
shap_importance_planet = pd.DataFrame({
    'Feature': X_train_planet.columns,
    'MeanAbsSHAP': mean_abs_shap_planet
})

# 3. 按贡献度降序排序
shap_importance_planet = shap_importance_planet.sort_values(
    by='MeanAbsSHAP', ascending=False
)

# 4. 计算相对贡献率（%）
shap_importance_planet['Contribution_%'] = (
    shap_importance_planet['MeanAbsSHAP'] /
    shap_importance_planet['MeanAbsSHAP'].sum() * 100
)

# 5. 取前 5 个变量
top5_shap_planet = shap_importance_planet.head(5)

print(top5_shap_planet)
```
